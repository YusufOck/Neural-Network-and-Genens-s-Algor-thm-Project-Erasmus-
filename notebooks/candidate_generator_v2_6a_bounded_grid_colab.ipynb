# Candidate Generator v2.6a — bounded same-leaf/channel grid

Bu defter mevcut v2.3 Drive artifact'larını **salt-okunur** kullanır. Üç önceden dondurulmuş Kidney/Lymphoma profilini karşılaştırır. Dataset ZIP'i gerekmez; duplicate scoring, family, split veya model eğitimi yapmaz.


```python
from pathlib import Path
import json, subprocess, sys
from google.colab import drive

drive.mount('/content/drive')
REPOSITORY_URL = 'https://github.com/YusufOck/Neural-Network-and-Genens-s-Algor-thm-Project-Erasmus-.git'
PROJECT_ROOT = Path('/content/NNGA')
if (PROJECT_ROOT / '.git').is_dir():
    subprocess.run(['git', '-C', str(PROJECT_ROOT), 'pull', '--ff-only'], check=True)
else:
    subprocess.run(['git', 'clone', REPOSITORY_URL, str(PROJECT_ROOT)], check=True)
commit = subprocess.check_output(['git', '-C', str(PROJECT_ROOT), 'rev-parse', 'HEAD'], text=True).strip()
OUTPUT_DIR = Path('/content/drive/MyDrive/NNGA_OUTPUTS/candidate_generator_v2_3_full_retrieval_v1')
REPORT = Path('/content/drive/MyDrive/NNGA_OUTPUTS/candidate_generator_v2_6a_bounded_grid_v1.json')
assert (OUTPUT_DIR / 'stage_manifest.json').is_file(), 'stage_manifest.json bulunamadı.'
assert (OUTPUT_DIR / 'run_contract.json').is_file(), 'run_contract.json bulunamadı.'
assert (PROJECT_ROOT / 'config/candidate_generator_v2_6a_bounded_grid.json').is_file()
assert (PROJECT_ROOT / 'config/candidate_generator_v2_6a_bounded_grid.sha256').is_file()
assert (PROJECT_ROOT / 'src/analyze_candidate_generator_v2_6a_bounded_grid.py').is_file()
print('Git commit:', commit)
print('V2.3 artifact klasörü hazır:', OUTPUT_DIR)
print('Dataset ZIP veya görüntü klasörü gerekmiyor.')
```

```python
command = [
    sys.executable, '-u',
    str(PROJECT_ROOT / 'src/analyze_candidate_generator_v2_6a_bounded_grid.py'),
    '--output-dir', str(OUTPUT_DIR),
    '--report', str(REPORT),
    '--device', 'cuda',
    '--block-size', '256',
]
print('Salt-okunur v2.6a bounded-grid tanısı başlıyor.')
print('Komut:', ' '.join(command))
completed = subprocess.run(command, cwd=PROJECT_ROOT)
print('RETURN CODE:', completed.returncode)
if completed.returncode != 0:
    raise RuntimeError('Tanı tamamlanamadı; eski artifactlar değiştirilmedi. Bu çıktıyı Codex’e gönder.')
payload = json.loads(REPORT.read_text(encoding='utf-8'))
print('Durum:', payload['status'])
for row in payload['profiles']:
    print(row['profile_id'], '| yeni=', row['new_same_leaf_pairs'], '| toplam=', row['full_candidate_pairs'], '| çift/görüntü=', round(row['pairs_per_image'], 6), '| hit=', row['development_same2d_hits'], '/4', '| kapasite=', row['candidate_load_passed'])
print('Seçilen geliştirme profili:', payload['selected_development_profile'])
print('Kapılar:', payload['acceptance_gates'])
print('Rapor:', REPORT)
```
