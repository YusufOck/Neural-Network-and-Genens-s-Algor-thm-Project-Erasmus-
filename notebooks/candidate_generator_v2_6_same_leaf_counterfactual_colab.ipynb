# Candidate Generator v2.6 — same-leaf counterfactual

Bu defter mevcut v2.3 Drive artifact'larını **salt-okunur** kullanır. Dataset ZIP'i gerekmez. Duplicate scoring, family, split veya model eğitimi yapmaz.


```python
from pathlib import Path
import json, subprocess, sys
from google.colab import drive

drive.mount('/content/drive')
REPOSITORY_URL = 'https://github.com/YusufOck/Neural-Network-and-Genens-s-Algor-thm-Project-Erasmus-.git'
PROJECT_ROOT = Path('/content/NNGA')
if (PROJECT_ROOT / '.git').is_dir():
    subprocess.run(['git', '-C', str(PROJECT_ROOT), 'pull', '--ff-only'], check=True)
else:
    subprocess.run(['git', 'clone', REPOSITORY_URL, str(PROJECT_ROOT)], check=True)
commit = subprocess.check_output(['git', '-C', str(PROJECT_ROOT), 'rev-parse', 'HEAD'], text=True).strip()
OUTPUT_DIR = Path('/content/drive/MyDrive/NNGA_OUTPUTS/candidate_generator_v2_3_full_retrieval_v1')
REPORT = Path('/content/drive/MyDrive/NNGA_OUTPUTS/candidate_generator_v2_6_same_leaf_counterfactual_v1.json')
assert (OUTPUT_DIR / 'stage_manifest.json').is_file(), 'stage_manifest.json bulunamadı.'
assert (OUTPUT_DIR / 'run_contract.json').is_file(), 'run_contract.json bulunamadı.'
assert (PROJECT_ROOT / 'config/candidate_generator_v2_6_same_leaf_counterfactual.json').is_file()
assert (PROJECT_ROOT / 'src/analyze_candidate_generator_v2_6_same_leaf_counterfactual.py').is_file()
print('Git commit:', commit)
print('V2.3 artifact klasörü hazır:', OUTPUT_DIR)
print('Dataset ZIP veya görüntü klasörü gerekmiyor.')
```

```python
command = [
    sys.executable, '-u',
    str(PROJECT_ROOT / 'src/analyze_candidate_generator_v2_6_same_leaf_counterfactual.py'),
    '--output-dir', str(OUTPUT_DIR),
    '--report', str(REPORT),
    '--device', 'cuda',
    '--block-size', '256',
]
print('Salt-okunur tanı başlıyor.')
print('Komut:', ' '.join(command))
completed = subprocess.run(command, cwd=PROJECT_ROOT)
print('RETURN CODE:', completed.returncode)
if completed.returncode != 0:
    raise RuntimeError('Tanı tamamlanamadı; eski artifactlar değiştirilmedi. Bu çıktıyı Codex’e gönder.')
payload = json.loads(REPORT.read_text(encoding='utf-8'))
print('Durum:', payload['status'])
print('Yeni çift:', payload['candidate_load']['new_same_leaf_pairs'])
print('Toplam çift:', payload['candidate_load']['proposed_full_pairs'])
print('Çift/görüntü:', payload['candidate_load']['pairs_per_image'])
print('Geliştirme same-2D:', payload['development_same2d_reference']['hits'], '/ 4')
print('Kapılar:', payload['acceptance_gates'])
print('Rapor:', REPORT)
```
